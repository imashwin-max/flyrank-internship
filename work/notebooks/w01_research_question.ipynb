## 1. My lane (or freestyle) and why

I have chosen the **Refresh / Content Opportunity Scoring** lane. This lane appeals to me because it is highly practical and directly actionable. Websites constantly accumulate stale content, and editorial teams have limited capacity. Instead of guessing which articles to update or just updating the oldest ones, this project aims to predict which pages are showing the strongest signals of decay but still hold significant traffic potential, allowing for a data-driven prioritization of editorial effort.

```python
import pandas as pd
import os

# Determine correct path to data
data_path = '../../data/raw/content_refresh_anonymized.csv'
if not os.path.exists(data_path):
    data_path = '../data/raw/content_refresh_anonymized.csv'
if not os.path.exists(data_path):
    data_path = '../../../flyrank-ml-internship-starter/data/raw/content_refresh_anonymized.csv'

df = pd.read_csv(data_path)
print(f"Dataset loaded with {df.shape[0]} pages.")
```

## 2. The question: decision, action, cost of a wrong call

- **Decision:** Which existing page should the content team rewrite or update next to defend or regain traffic?
- **Action:** An editor or content manager takes the top N pages from the model's ranked queue and assigns them to writers for a content refresh.
- **Cost of a wrong call:** Wasting editorial resources (time and budget) rewriting a page that doesn't actually need it or won't benefit from it, while a page that is actively losing valuable traffic continues to decay unaddressed.

```python
# Let's look at the distribution of trend directions
trend_counts = df['trend_direction'].value_counts(normalize=True) * 100
print("Percentage of pages by trend direction:")
print(trend_counts.round(1))
```

## 3. Quick look at the data (2-3 real numbers)

Looking at the starter dataset, we can see why this is a non-trivial problem worth solving with ML. For example, the data shows a significant proportion of pages are actively declining. However, distinguishing a 'valuable' declining page from a 'low-value' declining page requires looking at impressions and content age simultaneously. The median declining page has over 300 days since its last update and is still getting thousands of impressions, meaning there is real traffic at stake.

```python
declining_df = df[df['trend_direction'] == 'down']
growing_df = df[df['trend_direction'] == 'up']

median_impressions_down = declining_df['impressions_90d'].median()
median_age_down = declining_df['days_since_last_update'].median()

print(f"Median impressions (90d) for declining pages: {median_impressions_down:,.0f}")
print(f"Median days since last update for declining pages: {median_age_down:.0f} days")
print(f"Total declining pages with >10k impressions: {len(declining_df[declining_df['impressions_90d'] > 10000])}")
```

## 4. Careful words: what I can and can't claim

I **can** claim that the model observes historical patterns (such as impression volume, average position, and content staleness) and identifies pages that are statistically likely to be on a downward trend. The output is a **directional, decision-support tool** for prioritizing human review.

I **cannot** claim that updating a page will causally reverse its decline, nor can I claim that my model has 'reverse-engineered' how Google's algorithm decides what is fresh or relevant. This is an observational ranking tool, not a causal proof.
